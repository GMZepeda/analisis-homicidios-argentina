```python
import pandas as pd

# Cargamos las bases (recordá el sep=';' si el csv lo requiere)
df_homicidios = pd.read_csv('homicidios_limpio.csv', sep=';') 
df_censo = pd.read_excel('censo_limpio.xlsx')
```

```python
# Unimos por el nombre de la provincia
df_merged = df_homicidios.merge(df_censo, on='provincia_nombre', how='inner')
```

```python
# Contamos homicidios y traemos la población
df_provincias = df_merged.groupby('provincia_nombre').agg(
    total_casos=('id_hecho', 'count'),
    poblacion=('poblacion', 'first')
).reset_index()
```

```python
# Calculamos la tasa cada 100.000 habitantes
df_provincias['tasa_100k'] = (df_provincias['total_casos'] / df_provincias['poblacion']) * 100000
```

```python
# Ordenamos y mostramos las 3 mayores
top_3_provincias = df_provincias.sort_values(by='tasa_100k', ascending=False).head(3)
top_3_provincias[['provincia_nombre', 'tasa_100k', 'total_casos']]
```
