```python
df_censo = pd.read_excel('C:\\PROYECTOS\\SQL\\homicidios\\data\\c2022_tp_c_resumen.xlsx', skiprows=4, header=None)
```

```python
# Como le dijimos header=None, las columnas se llamarán con números (0, 1, 2...). 
# Por Excel, sabemos que:
# Columna 0 = Jurisdicción (Provincia)
# Columna 1 = Total de población

df_censo = df_censo.iloc[:, [0, 1]] # Nos quedamos solo con esas dos columnas
df_censo.columns = ['provincia_nombre', 'poblacion'] # Les ponemos nombre claro
```

```python
# Limpiamos posibles filas vacías al final o al principio
df_censo = df_censo.dropna(subset=['provincia_nombre'])
```

```python
#  Nos quedamos solo con las filas de las 24 jurisdicciones (excluimos notas y totales)
df_censo = df_censo.iloc[1:25].copy()

# Estandarizar los nombres de las provincias
df_censo['provincia_nombre'] = df_censo['provincia_nombre'].replace({
    'Tierra del Fuego, Antártida e Islas del Atlántico Sur': 'Tierra del Fuego',
    'Ciudad Autónoma de Buenos Aires': 'Ciudad Autonoma de Buenos Aires',
    'Córdoba': 'Cordoba',
    'Tucumán': 'Tucuman'
})
```
