```python
# imports necesarios
import pandas as pd
```

```python
# Cargamos el dataset CRUDO 
df_crudo = pd.read_csv('../data/SAT-HD-BU.csv', sep=',')
```

```python
# Mostrar los nombres de todas las columnas
print(df_crudo.columns.tolist())
```

```python
# .duplicated() marca con True las filas cuyo 'id_hecho' ya apareció antes, y .sum() cuenta cuántos True hay
duplicados = df_crudo.duplicated(subset=['id_hecho']).sum()

# Mostramos el total 
print("Registros con id_hecho duplicado:", duplicados)
```

**Los deuplicados de id_hecho son 23619**
