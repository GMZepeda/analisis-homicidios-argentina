```python
import pandas as pd
```

```python
df = pd.read_csv('involucrados.csv')
```

```python
# Separar Víctimas
df_victimas = df[['id_hecho',
'victima_sexo',
'victima_identidad_genero',
'victima_identidad_genero_otro',
'victima_tr_edad',
'victima_18_anios_o_mas']].copy()

df_victimas.rename(columns={
    'id_hecho': 'hecho_id',
    'victima_sexo': 'sexo',
    'victima_identidad_genero': 'identidad_genero',
    'victima_identidad_genero_otro': 'identidad_genero_otro',
    'victima_tr_edad': 'tr_edad', 
    'victima_18_anios_o_mas': 'mayor_18_anios'
}, inplace=True)

df_victimas['tipo_involucrado'] = 'victima'
```

```python
df_victimas.head()
```

```python
# Separar Inculpados
df_inculpados = df[['id_hecho',
'inculpado_sexo',
'inculpado_identidad_genero',
'inculpado_identidad_genero_otro',
'inculpado_tr_edad',
'inculpado_18_anios_o_mas']].copy()

df_inculpados.rename(columns={
    'id_hecho': 'hecho_id',
    'inculpado_sexo': 'sexo',
    'inculpado_identidad_genero': 'identidad_genero',
    'inculpado_identidad_genero_otro': 'identidad_genero_otro',
    'inculpado_tr_edad': 'tr_edad', 
    'inculpado_18_anios_o_mas': 'mayor_18_anios'
}, inplace=True)
df_inculpados['tipo_involucrado'] = 'inculpado'
```

```python
df_inculpados.head()
```

```python
# Unir ambos DataFrames
df_involucrados_melt = pd.concat([df_victimas, df_inculpados], ignore_index=True)
```

```python
# Exportar el nuevo CSV
df_involucrados_melt.to_csv('involucrados.csv', index=False, encoding='utf-8')
```

```python
# Ordenamos las columnas exactamente como las espera SQL
orden_columnas = [
    'tipo_involucrado', 
    'sexo', 
    'identidad_genero', 
    'identidad_genero_otro', 
    'tr_edad', 
    'mayor_18_anios', 
    'hecho_id'
]
df_involucrados_melt = df_involucrados_melt[orden_columnas]

# Exportamos el CSV limpio y ordenado
df_involucrados_melt.to_csv('involucrados.csv', index=False, encoding='utf-8')
```
