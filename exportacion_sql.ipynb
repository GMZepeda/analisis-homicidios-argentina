```python
import pandas as pd
```

```python
# cargar el archivo limpio
df = pd.read_csv('homicidios_limpio.csv', sep=';')
```

```python
df.columns
```

```python
# Tabla PROVINCIA
df_provincia = df[['provincia_id', 'provincia_nombre']].drop_duplicates().reset_index(drop=True)
```

```python
df_provincia.head()
```

```python
# Tabla LOCALIDAD
df_localidad = df[['localidad_id','localidad_nombre']].drop_duplicates().reset_index(drop=True)
```

```python
df_localidad.head()
```

```python
# Tabla HECHO
df_hecho = df[['id_hecho', 'localidad_id', 'anio', 'mes',
 'fecha_hecho', 'hora_hecho', 'tipo_lugar', 'tipo_lugar_otro',
  'tipo_lugar_ampliado', 'motivo_origen_registro',
   'motivo_origen_registro_otro']].drop_duplicates().reset_index(drop=True)
```

```python
df_hecho.head()
```

```python
# Tabla INVOLUCRADOS
columnas_involucrados = ['id_hecho',
'victima_sexo',
'victima_identidad_genero',
'victima_identidad_genero_otro',
'victima_tr_edad',
'victima_18_anios_o_mas',
'inculpado_sexo',
'inculpado_identidad_genero',
'inculpado_identidad_genero_otro',
'inculpado_tr_edad',
'inculpado_18_anios_o_mas']

df_involucrados = df[columnas_involucrados].drop_duplicates().reset_index(drop=True)
```

```python
# Exportamos las tablas
df_provincia.to_csv('provincia.csv', index=False, encoding='utf-8')
df_localidad.to_csv('localidad.csv', index=False, encoding='utf-8')
df_hecho.to_csv('hecho.csv', index=False, encoding='utf-8')
#df_involucrados.to_csv('involucrados.csv', index=False, encoding='utf-8')
```
